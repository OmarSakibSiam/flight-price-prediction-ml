**Feature Engineering**

```python
# Handling Date and time column
# Date of Journy
df['Date']=df['Date_of_Journey'].str.split('/').str[0].astype(int)
df['Month']=df['Date_of_Journey'].str.split('/').str[1].astype(int)
df['Year']=df['Date_of_Journey'].str.split('/').str[2].astype(int)
```

```python
df.info()
```

```python
df.drop('Date_of_Journey',axis=1,inplace=True)
```

```python
# Handling Arrival Time
df['Arrival_Time']=df['Arrival_Time'].str.split(' ').str[0]
```

```python
df['Arrival_Time_hours']=df['Arrival_Time'].str.split(':').str[0].astype(int)
df['Arrival_Time_minutes']=df['Arrival_Time'].str.split(':').str[1].astype(int)
```

```python
df.drop('Arrival_Time',axis=1,inplace=True)
```

```python
# Handling Departure Time
df['Dep_Time_hours']=df['Dep_Time'].str.split(':').str[0].astype(int)
df['Dep_Time_minutes']=df['Dep_Time'].str.split(':').str[1].astype(int)
```

```python
df.drop('Dep_Time',axis=1,inplace=True)
```

```python
df_train=df[~df['Price'].isnull()]
df_test=df[df['Price'].isnull()]
```

```python
df_test.head()
```

```python
X=df_train.drop('Price',axis=1)
y=df_train['Price']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
# Random Forest Regressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score

rf = RandomForestRegressor(n_estimators=200, max_depth=None, min_samples_leaf=2,
                            random_state=42, n_jobs=-1)
rf.fit(X_train, y_train)

y_pred_rf = rf.predict(X_test)
score_rf = r2_score(y_test, y_pred_rf)
print(f"R² Score (Random Forest): {score_rf:.3f}")
```

```python
X_unknown = df_test.drop('Price', axis=1)

predicted_prices = rf.predict(X_unknown)
```
